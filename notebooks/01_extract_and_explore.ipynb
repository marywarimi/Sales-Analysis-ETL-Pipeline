```python
import pandas as pd

df = pd.read_csv("../data/raw/superstore.csv", encoding="latin-1")

df.head()
```

```python
print(df.shape)

df.info()
```

```python
print("Duplicate rows:", df.duplicated().sum())

df["Order Date"] = pd.to_datetime(df["Order Date"], format="%m/%d/%Y")
df["Ship Date"] = pd.to_datetime(df["Ship Date"], format="%m/%d/%Y")

df["Postal Code"] = df["Postal Code"].astype(str).str.zfill(5)

df[["Order Date", "Ship Date", "Postal Code"]].info()
```

```python
print(df[["Sales", "Quantity", "Discount", "Profit"]].describe())

print("Ship before order:", (df["Ship Date"] < df["Order Date"]).sum())

df["Order Year"] = df["Order Date"].dt.year
df["Order Month"] = df["Order Date"].dt.month
df["Shipping Days"] = (df["Ship Date"] - df["Order Date"]).dt.days

df[["Order Date", "Order Year", "Order Month", "Shipping Days"]].head()
```
