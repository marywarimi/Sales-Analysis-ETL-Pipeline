```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("../data/superstore.db")

pd.read_sql("SELECT * FROM sales LIMIT 5", conn)
```

```python
pd.read_sql(
    """
    SELECT region,
           ROUND(SUM(sales), 2) AS total_sales,
           ROUND(SUM(profit), 2) AS total_profit
    FROM sales
    GROUP BY region
    ORDER BY total_sales DESC
    """,
    conn,
)
```

```python
pd.read_sql(
    """
    SELECT sub_category,
           ROUND(SUM(sales), 2) AS total_sales,
           ROUND(SUM(profit), 2) AS total_profit,
           ROUND(AVG(discount), 2) AS avg_discount
    FROM sales
    GROUP BY sub_category
    ORDER BY total_profit ASC
    """,
    conn,
)
```

```python
pd.read_sql(
    """
    SELECT discount,
           COUNT(*) AS items_sold,
           ROUND(SUM(profit), 2) AS total_profit
    FROM sales
    GROUP BY discount
    ORDER BY discount
    """,
    conn,
)
```

```python
pd.read_sql(
    """
    SELECT discount,
           COUNT(*) AS items_sold,
           ROUND(SUM(profit), 2) AS total_profit
    FROM sales
    WHERE sub_category = 'Tables'
    GROUP BY discount
    ORDER BY discount
    """,
    conn,
)
```

# Sales Analysis: Key Findings

Data: Sample Superstore (9,994 items sold, 2014 to 2017), cleaned with pandas and queried with SQL in SQLite.

## What I found

- **Sales are growing.** Yearly sales went from about 484K in 2014 to about 733K in 2017 (a small dip in 2015).
- **West is the best region.** It has the highest sales (about 725K) and the highest profit (about 108K).
- **Central sells a lot but keeps little.** It earned only about 8 profit for every 100 sold, the lowest of all regions.
- **Tables lose money.** They sold about 207K but lost about 17.7K, with an average discount of 26%.
- **Discounts above 20% lose money.** Items sold at 30% off or more (about 1,393 items) lost about 135K in total, while items at 0 to 20% off made about 421K.
